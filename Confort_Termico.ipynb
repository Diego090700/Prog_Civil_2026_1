## **Teoría del Confort Térmico en Ganado Vacuno**

El confort térmico en el ganado vacuno es un estado en el que el animal no experimenta estrés por calor ni por frío, lo que permite mantener su temperatura corporal dentro de rangos normales sin necesidad de activar mecanismos fisiológicos o conductuales excesivos. Este estado es crucial para la producción de leche y carne, así como para la salud y el bienestar animal.

### **Estrés por Calor**

El estrés por calor es una de las principales preocupaciones en la producción ganadera, especialmente en regiones cálidas o durante los meses de verano. Se produce cuando la capacidad del animal para disipar el calor (generado internamente por el metabolismo y absorbido del ambiente) es superada. Los factores que contribuyen al estrés por calor incluyen:

*   **Temperatura Ambiente:** Temperaturas elevadas dificultan la disipación de calor por convección y radiación.
*   **Humedad Relativa:** Una alta humedad reduce la eficiencia de la evaporación (sudoración y jadeo), que es un mecanismo clave para la pérdida de calor.
*   **Radiación Solar:** La exposición directa al sol aumenta la carga térmica.
*   **Velocidad del Viento:** El viento puede ayudar a la disipación de calor, pero su efecto disminuye con altas temperaturas.

### **Indicadores de Estrés por Calor**

El Índice de Temperatura y Humedad (ITH) es una herramienta ampliamente utilizada para evaluar el nivel de estrés por calor en el ganado. Combina la temperatura de bulbo seco y la temperatura de bulbo húmedo para proporcionar una medida de la carga térmica ambiental. Otros indicadores incluyen:

*   **Frecuencia Respiratoria:** Aumento del jadeo para disipar calor por evaporación.
*   **Temperatura Rectal:** Elevación de la temperatura corporal.
*   **Consumo de Agua:** Incremento en la ingesta de agua.
*   **Consumo de Alimento:** Reducción del consumo de materia seca.
*   **Producción:** Disminución en la producción de leche y/o carne, y reducción de la fertilidad.

### **Manejo del Confort Térmico**

Para mitigar el estrés por calor y asegurar el confort térmico, se implementan diversas estrategias:

*   **Sombra:** Provisión de sombra natural o artificial para reducir la carga de radiación solar.
*   **Ventilación:** Uso de ventiladores para aumentar el flujo de aire y facilitar la disipación de calor.
*   **Aspersores/Nebulizadores:** Sistemas de enfriamiento evaporativo que rocían agua fina sobre los animales.
*   **Acceso a Agua Fresca:** Asegurar disponibilidad constante de agua fresca y limpia.
*   **Manejo Nutricional:** Ajustes en la dieta para reducir la generación de calor metabólico y proporcionar suplementos que ayuden a mitigar los efectos del estrés.

El monitoreo continuo del ITH y otros indicadores es fundamental para tomar decisiones oportunas y mantener al ganado en un estado de confort térmico óptimo, lo que se traduce en mayor productividad y bienestar.

# **Metodos y Calculos**
* indice de temperatura y humeda (producción intensiva)

$TH = 0.72(T{bs} + T{bh})+40.6 $

donde:
1. $ T _ {bs}$ Temperatura bulto seco ($°C$)
2. $T_{bh}$ Temperatura del bulto humedo ($°C$)

Para la correlación de los datos de la especie bovina en la producción de leche se recomienda utilizar los límites propuestos por National Weather Service:

1. Si $ITH <= 70$ ambiente no estresante
2. S1 SITH > 70 \space and space ITH <= 78$
situación de alerta por estrés térmico
3. Si ITH 78 0 ≤ 83 peligro por estrés térmico
4. S1 ITH 83 > situación critica




```python
# ENTRADA: pedir tbs (Temperatura de Bulbo seco) y tbh (Temperatura de bulbo hümedo)
tbs = float(input("Ingrese la temperatura de bulbo seco, Grados Centigrados: "))
tbh = float(input("Ingrese la temperatura de bulbo húmedo, Grados Centigrados: "))

# PROCESO: Calculo de ith (Indice de Temperatura y Humedad
ith = 0.72 * (tbs + tbh) + 40.6

# SALIDA: Imprimir resultado, de la evaluación del ith y el confort térmico
if ith <= 70:
  print("El ambiente es no estresante")
elif ith > 70 and ith <= 78:
  print("La situación es de alerta por estrés térmico")
elif ith > 78 and ith <= 83:
  print("La situación es de  peligro por estrés térmico")
else:
  print("La situación es critica")
```

### **Explicación del Código de Confort Térmico**

Este código calcula el Índice de Temperatura y Humedad (ITH) y evalúa el nivel de estrés térmico en el ganado bovino basándose en los valores de temperatura de bulbo seco y bulbo húmedo.

Aquí la explicación de cada parte:

1.  **Entrada de Datos:**
    *   `tbs = float(input("Ingrese la temperatura de bulbo seco, Grados Centigrados: "))`:
        Esta línea solicita al usuario que introduzca la temperatura de bulbo seco en grados Celsius. La función `input()` lee la entrada del usuario como una cadena de texto, y `float()` la convierte a un número de punto flotante para poder realizar cálculos con ella. El valor se almacena en la variable `tbs`.
    *   `tbh = float(input("Ingrese la temperatura de bulbo húmedo, Grados Centigrados: "))`:
        Similar a la anterior, esta línea solicita y captura la temperatura de bulbo húmedo del usuario, la convierte a un número de punto flotante y la almacena en la variable `tbh`.

2.  **Cálculo del ITH:**
    *   `ith = 0.72 * (tbs + tbh) + 40.6`:
        Esta es la fórmula principal para calcular el Índice de Temperatura y Humedad (ITH). Se multiplica la suma de la temperatura de bulbo seco (`tbs`) y la temperatura de bulbo húmedo (`tbh`) por `0.72`, y luego se le suma `40.6`. El resultado de este cálculo se guarda en la variable `ith`.

3.  **Evaluación del Confort Térmico:**
    *   `if ith <= 70:`:
        Esta es la primera condición. Si el valor del ITH es menor o igual a `70`, significa que el ambiente no es estresante.
    *   `print("El ambiente es no estresante")`:
        Si la condición anterior es verdadera, se imprime este mensaje indicando que no hay estrés térmico.
    *   `elif ith > 70 and ith <= 78:`:
        Si la primera condición (`ith <= 70`) es falsa, el programa pasa a evaluar esta segunda condición. Si el ITH es mayor que `70` pero menor o igual que `78`, se considera una situación de alerta por estrés térmico.
    *   `print("La situación es de alerta por estrés térmico")`:
        Si la segunda condición es verdadera, se imprime este mensaje.
    *   `elif ith > 78 and ith <= 83:`:
        Si las dos condiciones anteriores son falsas, se evalúa esta tercera. Si el ITH es mayor que `78` pero menor o igual que `83`, indica un peligro por estrés térmico.
    *   `print("La situación es de peligro por estrés térmico")`:
        Si la tercera condición es verdadera, se imprime este mensaje.
    *   `else:`:
        Si ninguna de las condiciones anteriores es verdadera (es decir, el ITH es mayor que `83`), el programa ejecuta el bloque de código dentro de `else`.
    *   `print("La situación es critica")`:
        Si el ITH es mayor que `83`, se imprime este mensaje, indicando una situación crítica de estrés térmico.
